```python
pelamar = [("Budi", 3.2, 5),
           ("Siti", 3.6, 3), 
           ("Dodi", 3.8, 1)]
print("=== STATUS SELEKSI PELAMAR ===")
for nama, ipk, semester in pelamar:
    if ipk >= 3.5 and semester > 2:
        status = "DITERIMA"
    else:
        status = "TIDAK DITERIMA"
    print(f" Nama Pelamar: {nama:<6} | (Status: {status})")
```

```python
karyawan = [("Joko", "IT", 7000), 
            ("Rina", "HR", 5000), 
            ("Toni", "Data", 8000), 
            ("Lia", "IT", 6500)]
total_gaji=0
for nama, departemen, harga in karyawan:
    if departemen == "IT" or departemen == "Data":
        total_gaji += harga

print(f"Total Gaji untuk karyawan departemen IT dan Data adalah: Rp{total_gaji:,}")
```

```python
penjualan = [("Kopi", 60, False), 
             ("Teh", 30, False), 
             ("Boba", 80, True)]

for minuman, jumlah_terjual, sedang_promo in penjualan: 
    if  jumlah_terjual > 50 and not sedang_promo:
        print(f"Minuman yang laku keras adalah {minuman}")
```
